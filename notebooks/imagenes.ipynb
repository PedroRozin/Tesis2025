```python
import matplotlib.pyplot as plt
import numpy as np

# Configuración de la figura
fig, ax = plt.subplots(figsize=(12, 7))

# Parámetros geométricos (unidades arbitrarias para replicar la forma)
eta_hoy = 10
eta_desacople = 2.5
r_interseccion = eta_hoy - eta_desacople  # Donde el cono cruza el desacople
r_horizonte = r_interseccion + eta_desacople # Extensión del horizonte de partícula en t=desacople

# --- 1. DIBUJO DE LOS CONOS Y EL FONDO ---

# Área sombreada (Era de desacople/recombinación)
ax.fill_between([-15, 15], 0, eta_desacople, color='lightgray', alpha=0.3, zorder=0)

# Cono de luz pasado (Principal - Amarillo)
# Líneas desde (0, eta_hoy) hasta la intersección
ax.plot([0, r_interseccion], [eta_hoy, eta_desacople], color='#FDC54E', linewidth=4, zorder=2)
ax.plot([0, -r_interseccion], [eta_hoy, eta_desacople], color='#FDC54E', linewidth=4, zorder=2)

# Conos pequeños (Horizontes pasados en el desacople)
# Triángulo derecho
ax.plot([r_interseccion, r_horizonte], [eta_desacople, 0], color='#FDC54E', linewidth=4, zorder=2)
ax.plot([r_interseccion, r_interseccion - eta_desacople], [eta_desacople, 0], color='#FDC54E', linewidth=4, zorder=2)
# Triángulo izquierdo
ax.plot([-r_interseccion, -r_horizonte], [eta_desacople, 0], color='#FDC54E', linewidth=4, zorder=2)
ax.plot([-r_interseccion, -(r_interseccion - eta_desacople)], [eta_desacople, 0], color='#FDC54E', linewidth=4, zorder=2)

# --- 2. MODIFICACIONES SOLICITADAS ---

# A) Línea punteada horizontal negra en eta_desacople
ax.hlines(y=eta_desacople, xmin=-15, xmax=15, colors='black', linestyles='--', linewidth=1)
ax.text(0.5, eta_desacople + 0.3, r'$\eta_{desacople}$', fontsize=14, ha='left')

# B) Línea punteada vertical naranja (Intersección -> Horizonte Luminoso)
# La dibujamos solo en el lado derecho para no saturar, o en ambos si prefieres. Lo haré en el derecho como referencia.
ax.vlines(x=r_interseccion, ymin=0, ymax=eta_desacople, colors='orange', linestyles='--', linewidth=1.5)
# Label 'Horizonte Luminoso' (con flecha para mayor claridad)
ax.annotate('Horizonte Luminoso', xy=(r_interseccion, eta_desacople/2), xytext=(r_interseccion + 1, eta_desacople + 1.5),
            arrowprops=dict(arrowstyle="->", color='orange'), fontsize=12, color='darkorange', ha='center')

# C) Línea punteada vertical azul (Horizonte -> Horizonte de partícula)
# Esta va en la base de los conos pequeños (donde dice 'r horizonte' en tu mente)
ax.vlines(x=r_horizonte, ymin=0, ymax=eta_desacople/2, colors='blue', linestyles='--', linewidth=1.5)
ax.text(r_horizonte, -0.8, 'Horizonte de\npartícula', fontsize=12, color='blue', ha='center', va='top')

# --- 3. EJES Y ETIQUETAS ---

# Configurar ejes centrados
ax.spines['left'].set_position('center')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# Flechas en los ejes
ax.plot(0, eta_hoy + 1, "^k", transform=ax.get_yaxis_transform(), clip_on=False) # Flecha Y
ax.plot(1, 0, ">k", transform=ax.get_xaxis_transform(), clip_on=False) # Flecha X der
ax.plot(0, 0, "<k", transform=ax.get_xaxis_transform(), clip_on=False) # Flecha X izq

# Etiquetas de Ejes
ax.set_ylabel(r'$\eta$', fontsize=16, rotation=0, loc='top')
# CAMBIO SOLICITADO: Letra 'r' en lugar de 'chi'
ax.text(14, 0.2, r'$r, (\theta=0)$', fontsize=16)
ax.text(-15, 0.2, r'$(\theta=\pi)$', fontsize=16)

# Etiquetas de puntos clave
ax.text(0.2, eta_hoy, r'$\eta_{hoy}$', fontsize=14)

# Etiquetas r_horizonte en el eje X
ax.text(r_horizonte, 0.2, r'$r_{horizonte}$', fontsize=12, ha='center')
ax.text(-r_horizonte, 0.2, r'$r_{horizonte}$', fontsize=12, ha='center')

# Limites y limpieza
ax.set_xlim(-15, 15)
ax.set_ylim(-1, 12)
ax.set_xticks([]) # Quitar números del eje X
ax.set_yticks([]) # Quitar números del eje Y

plt.tight_layout()
plt.show()
```

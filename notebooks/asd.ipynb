```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Arc

# Configuración
plt.figure(figsize=(10, 6))
ax = plt.gca()

# Parámetros
N = 8  # Número total de fuentes (solo para cálculo, mostraremos menos)
d = 1
b = (N)*d
D = 15
theta = np.radians(15)

# Coordenadas de las fuentes (en eje Y ahora)
y_sources = np.linspace(0, b, N, endpoint=False) 
x_sources = np.zeros(N)

# Dibujar todas las fuentes (puntos pequeños)
plt.scatter(x_sources, y_sources, color='blue', s=10, zorder=3)

# Marcar fuentes específicas con círculos más grandes y etiquetas
fuentes_especiales = [0, N//2, N-1]
for i in fuentes_especiales:
    plt.scatter(x_sources[i], y_sources[i], color='blue', s=80, zorder=4)
    if i == 0:
        plt.text(x_sources[i]-0.5, y_sources[i], f'$S_1$', va='center', ha='right')
    elif i == N//2:
        plt.text(x_sources[i]-0.5, y_sources[i], f'$S_{{N/2}}$', va='center', ha='right')
    else:
        plt.text(x_sources[i]-0.5, y_sources[i], f'$S_N$', va='center', ha='right')

# Línea central punteada
y_center = b/2
plt.plot([0, D], [y_center, y_center], 'k--', lw=1)

# Pantalla
plt.plot([D, D], [-1, b+1], 'k-', lw=3)
plt.text(D, b+1.5, 'Pantalla', ha='center')

# Punto P
y_p = y_center + D*np.tan(theta)
plt.scatter(D, y_p, color='red', s=80, zorder=3)
plt.text(D+0.3, y_p, 'P', va='center')

# Rayos
plt.plot([0, D], [y_center, y_p], 'k-', lw=1.5)  # Rayo central
plt.plot([0, D], [y_sources[0], y_p], 'gray', linestyle=':', lw=1)  # Rayo desde fuente 1
plt.text(1, y_center-3.2, r'$r_1$', color='gray', fontsize=14)
plt.text(2.2, y_center+1, r'$r_{N/2}$', color='k', fontsize=14)

# Ángulo theta
arc_radius = 2
arc = Arc((0, y_center), arc_radius, arc_radius, theta1=0, theta2=np.degrees(theta), color='red')
ax.add_patch(arc)
plt.text(1, y_center+0.5, r'$\theta$', color='red', fontsize=14)

# Distancia d
plt.annotate('', xy=(-0.5, y_sources[0]), xytext=(-0.5, y_sources[1]),
             arrowprops=dict(arrowstyle='<->', color='green'))
plt.text(-0.8, (y_sources[0]+y_sources[1])/2, r'$d$', va='center', color='green', fontsize=12)


# Distancia b
plt.annotate('', xy=(-1, 0), xytext=(-1, b-1),
             arrowprops=dict(arrowstyle='<->', color='purple'))
plt.text(-1.5, b/2, r'$b = Nd$', va='center', color='purple', fontsize=12, rotation=90)

# Distancia D (horizontal)
plt.annotate('', xy=(0, -0.8), xytext=(D, -0.8),
             arrowprops=dict(arrowstyle='<->', color='orange'))
plt.text(D/2, -1.3, r'$D$', ha='center', color='orange', fontsize=12)

# Distancia y (vertical en pantalla)
# plt.plot([D-0.2, D-0.2], [y_center, y_p], 'orange', lw=1.5)
plt.annotate('', xy=(D+0.5, y_center), xytext=(D+0.5, y_p),
             arrowprops=dict(arrowstyle='<->', color='orange'))
plt.text(D+0.8, (y_center+y_p)/2, r'$y$', va='center', color='orange', fontsize=12)

# Configuración final
plt.axis('equal')
plt.xlim(-2, D+1)
plt.ylim(-1, b+2)
plt.axis('off')  # Quitar ejes
plt.tight_layout()
plt.show()
```
